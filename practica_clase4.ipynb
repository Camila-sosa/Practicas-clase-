```python
texto = "Hola Mundo"
texto2 = "Ingenieria Industrial"
texto3 = "Python"

print(texto)
print(texto2)
print(texto3)
```

```python
dato1 = b"Hola"
dato2 = bytearray(b"Python")

print(dato1)
print(dato2)
print(type(dato1))
print(type(dato2))
```

```python
lista = [10, 20, 30, "Hola", True]
tupla = (1, 2, 3, "Python", False)

print(lista)
print(tupla)
print(type(lista))
print(type(tupla))
```
