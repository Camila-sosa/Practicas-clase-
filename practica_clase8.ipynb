```python
calificacion = int(input("Ingresa tu calificación: "))

if calificacion >= 90:
    print("Excelente")
elif calificacion >= 70:
    print("Aprobado")
else:
    print("Reprobado")
```

```python
suma = 0

while suma < 3:
    entrada = input("Escribe una palabra: ")

    if entrada == "despedida":
        break

    suma += 1
    print("Intento", suma)

print("Programa terminado")
```
